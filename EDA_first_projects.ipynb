```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('/content/unclean_employee_eda_practice.csv',encoding='latin1')
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
# Convert 'Age' column to numeric, converting non-numeric invalid entries to NaN
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')

# Fill missing values with the mean value (a scalar)
df['Age'] = df['Age'].fillna(df['Age'].mean())
```

```python
df.isnull().sum()
```

```python

df['Salary'] = df['Salary'].fillna(int(df['Salary'].mean()))
```

```python
df
```

```python
df['Joining_Date'] = pd.to_datetime(df['Joining_Date'], errors='coerce')
```

```python
df.info()
```

```python
df = df.dropna(subset=['Joining_Date'])
```

```python
df.isnull().sum()
```

```python
df.columns
```

```python
df=df.dropna(subset=['City'])
```

```python
df.info()
```

```python
df['Performance_Score']=df['Performance_Score'].fillna(df['Performance_Score'].mean())
```

```python
df=df.dropna(subset='Email')
```

```python
df.info()
```

```python
df.duplicated().sum()
```

```python
df
```

```python
df['Age']=df['Age'].astype(int)
```

```python
df
```

```python
df['Experience'] = (pd.Timestamp.today() - df['Joining_Date']).dt.days
```

```python
df
```

```python
department=df['Department'].value_counts()
```

```python
sns.barplot(
    x=department.index,
    y=department.values
)
plt.title("Department Distribution")
plt.xlabel("Department")
plt.ylabel("Count")
```

IT and Finance Department employe Distribution high.


```python
sns.histplot(df['Age'],bins=8)
```

Most employees are around 27–28 years old

```python
sns.countplot(x='City', data=df)
plt.title("Employee Count by City")
plt.xlabel("City")
plt.ylabel("Count")
```

Delhi has the highest employee count among all cities in this dataset.


```python
sns.boxplot(x=df['Salary'])
plt.title('Salary Outliers')
plt.show()
```

```python
salary_by_dept = df.groupby('Department')['Salary'].mean()

sns.barplot(x=salary_by_dept.index, y=salary_by_dept.values)
plt.title('Average Salary by Department')
plt.xlabel('Department')
plt.ylabel('Average Salary')
plt.show()
```

So, IT Department average salary high.


```python
sns.scatterplot(data=df, x='Experience', y='Performance_Score')
plt.show()
```

```python
correlation = df['Performance_Score'].corr(df['Experience'])
print(correlation)
```

```python
print(round(correlation, 2))
```

Experience does not have a strong relationship with Performance Score in this dataset

```python
sns.scatterplot(data=df, x='Salary', y='Experience')
plt.show()
```

```python
correlation = df['Salary'].corr(df['Experience'])
print(round(correlation, 2))
```

Employees with more experience tend to have higher salaries.

conclusion:--
   The EDA shows that salary is moderately related to experience, while experience has little relationship with performance score. The IT department has relatively higher salaries, and the employee population is mainly concentrated around the 27–28 age range.
